```python
from kivy.app import App
from kivy.uix.label import Label
from kivy.clock import mainthread
from plyer import gps

class LocationApp(App):
    def build(self):
        # UI: A simple label to display coordinates
        self.display = Label(text="Waiting for GPS data...", font_size='20sp')
        return self.display

    def on_start(self):
        # Configure and start the GPS
        try:
            gps.configure(on_location=self.on_location, on_status=self.on_status)
            gps.start(minTime=1000, minDistance=1) # Update every 1 second or 1 meter
        except NotImplementedError:
            self.display.text = "GPS is not implemented for your platform"

    @mainthread
    def on_location(self, **kwargs):
        # This function runs every time the location changes
        lat = kwargs.get('lat')
        lon = kwargs.get('lon')
        self.display.text = f"Latitude: {lat}\nLongitude: {lon}"

    def on_status(self, stype, status):
        # Optional: Print status messages (e.g., GPS ON/OFF)
        print(f"GPS Status: {stype} - {status}")

    def on_stop(self):
        # Stop GPS when the app is closed to save battery
        gps.stop()

if __name__ == '__main__':
    LocationApp().run()
```
