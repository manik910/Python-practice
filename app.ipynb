```python
!pip install phonenumbers

import phonenumbers
from phonenumbers import geocoder, carrier

# Enter the phone number with country code (e.g., +1 for USA, +91 for India)
number = "+919182738998"
parsed_number = phonenumbers.parse(number)

# Get Country Name
location = geocoder.description_for_number(parsed_number, "en")
print(f"Location: {location}")

# Get Service Provider (Carrier)
service_provider = carrier.name_for_number(parsed_number, "en")
print(f"Service Provider: {service_provider}")
```

```python
!pip install opencage
from opencage.geocoder import OpenCageGeocode
import folium

key = 'YOUR_OPENCAGE_API_KEY'
geocoder_api = OpenCageGeocode(key)

# Get Latitude and Longitude of the region
results = geocoder_api.geocode(location)
lat = results[0]['geometry']['lat']
lng = results[0]['geometry']['lng']

print(f"Latitude: {lat}, Longitude: {lng}")

# Create an HTML map
my_map = folium.Map(location=[lat, lng], zoom_start=9)
folium.Marker([lat, lng], popup=location).add_to(my_map)

# Save the map
my_map.save("mylocation.html")
print("Map has been saved as 'mylocation.html'.")
```

```python
from kivy.app import App
from kivy.uix.label import Label
from kivy.clock import mainthread
from plyer import gps

class LocationApp(App):
    def build(self):
        # UI: A simple label to display coordinates
        self.display = Label(text="Waiting for GPS data...", font_size='20sp')
        return self.display

    def on_start(self):
        # Configure and start the GPS
        try:
            gps.configure(on_location=self.on_location, on_status=self.on_status)
            gps.start(minTime=1000, minDistance=1) # Update every 1 second or 1 meter
        except NotImplementedError:
            self.display.text = "GPS is not implemented for your platform"

    @mainthread
    def on_location(self, **kwargs):
        # This function runs every time the location changes
        lat = kwargs.get('lat')
        lon = kwargs.get('lon')
        self.display.text = f"Latitude: {lat}\nLongitude: {lon}"

    def on_status(self, stype, status):
        # Optional: Print status messages (e.g., GPS ON/OFF)
        print(f"GPS Status: {stype} - {status}")

    def on_stop(self):
        # Stop GPS when the app is closed to save battery
        gps.stop()

if __name__ == '__main__':
    LocationApp().run()
```
