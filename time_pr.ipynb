```python
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt

d = {
 "area": [1, 2, 3, 4],
 "rice_yield": [10, 20, 30, 40]     # to find the r values
}

df = pd.DataFrame(d)

X = df.area.values
y = df.rice_yield.values

plt.xlabel("X")
plt.ylabel("y")

plt.scatter(X, y)
plt.show()

slope, intercept, r_value, p_value, std_err = stats.linregress(X, y)
```

```python
import torch
import torch.nn.functional as F
q=torch.rand(1,3)
v=torch.rand(3,3)
k=torch.rand(3,3)

# Attention score
scores = torch.matmul(q, k.T)

# Softmax
weights = F.softmax(scores, dim=-1)

# Output
attention_output = torch.matmul(weights, v)

print(attention_output)
```

```python
import torch
import torch.nn.functional as F

Q=torch.rand(3,5,4)
K=torch.rand(3,5,4)
V=torch.rand(3,5,4)
scores=torch.matmul(Q,K.transpose(-2,-1))
weights=F.softmax(scores,dim=-1)
attention_output=torch.matmul(weights,V)
print(attention_output)
```

```python
n=13
for i in range(1,n+1):
  for j in range(1,n+1):
    top = i <= n/2
    bottom = i > n/2
    left = j<=i
    right = j<=n-i+1
    if (top and left and right) or (bottom and left and right):
      print("*", end="")
    else:
      print("",end=" ")
  print()
```
